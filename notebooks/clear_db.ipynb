# Limpeza das tabelas do ETL

Este notebook remove **todos os dados** das tabelas usadas pelo ETL de exercícios.

A operação é destrutiva. Faça um backup antes de executar. A confirmação é obrigatória e a limpeza usa uma única transação para permitir rollback em caso de erro.

```python
from sqlalchemy import create_engine, text


user='root'
password='*'
host='localhost'
port=3306
database='spring_api_system'

MYSQL_URL = f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}"

if not MYSQL_URL:
    raise RuntimeError(
        'Defina a variável de ambiente MYSQL_URL antes de executar a limpeza.'
    )

engine = create_engine(MYSQL_URL)
```

```python
def clear_database(confirm=False):
    """Remove todos os dados das tabelas do ETL em uma única transação."""
    if confirm is not True:
        raise ValueError(
            'Operação cancelada. Execute clear_database(confirm=True) somente após confirmar.'
        )

    tables = [
        'exercise_secondary_muscle',
        'exercise',
        'secondary_muscle',
        'equipment',
        'body_part',
    ]
    deleted_rows = {}

    try:
        with engine.begin() as connection:
            for table in tables:
                result = connection.execute(text(f'DELETE FROM {table}'))
                deleted_rows[table] = result.rowcount
    except Exception:
        print('Falha na limpeza. A transação foi revertida.')
        raise

    print('Limpeza concluída com sucesso:')
    for table, count in deleted_rows.items():
        print(f'- {table}: {count} linhas removidas')

    return deleted_rows
```

```python
deleted_rows = clear_database(confirm=True)
```
