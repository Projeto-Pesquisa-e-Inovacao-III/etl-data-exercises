# Remoção das tabelas do ETL

Este notebook remove as tabelas do ETL para permitir a recriação do schema desde o início.

A tabela associativa é removida primeiro porque possui chaves estrangeiras para as tabelas principais. Não é usado `CASCADE`: a ordem de remoção fica explícita e previsível.

**Atenção:** esta operação apaga as tabelas e todos os dados. Execute somente depois de confirmar que o banco correto está configurado e que existe backup, se necessário.


```python
import os

from sqlalchemy import create_engine, text
user='root'
password='*'
host='localhost'
port=3306
database='spring_api_system'


MYSQL_URL = f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}"

if not MYSQL_URL:
    raise RuntimeError(
        "Defina a variável de ambiente MYSQL_URL antes de remover as tabelas. "
        "Exemplo: mysql+pymysql://usuario:senha@localhost:3306/spring_api_system"
    )

engine = create_engine(MYSQL_URL)
```

```python
def drop_tables(confirm=False):
    """Remove todas as tabelas do ETL em uma única transação."""
    if confirm is not True:
        raise ValueError(
            "Operação cancelada. Execute drop_tables(confirm=True) somente "
            "após confirmar o banco de destino."
        )

    statements = [
        "DROP TABLE IF EXISTS exercise_secondary_muscle",
        "DROP TABLE IF EXISTS exercise",
        "DROP TABLE IF EXISTS secondary_muscle",
        "DROP TABLE IF EXISTS equipment",
        "DROP TABLE IF EXISTS body_part",
    ]

    try:
        with engine.begin() as connection:
            for statement in statements:
                connection.execute(text(statement))
    except Exception:
        print("Falha ao remover as tabelas. A transação foi revertida.")
        raise

    print(
        "Tabelas removidas: exercise_secondary_muscle, exercise, "
        "secondary_muscle, equipment e body_part."
    )
    print(
        "Agora execute notebooks/create_tables.ipynb para recriá-las "
        "com IDs AUTO_INCREMENT."
    )
```

```python
# Execute somente após confirmar o banco configurado:
drop_tables(confirm=True)
```
